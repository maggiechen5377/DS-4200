```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

spotify = pd.read_csv("spotify2000.csv")

artist_counts = spotify['artist_name'].value_counts().head(15)
mean_val = artist_counts.mean()

colors = ['#e05c2a' if i < 3 else 'steelblue' for i in range(len(artist_counts))]
colors = colors[::-1]  # flip to match bar order

plt.figure(figsize=(10, 6))
plt.barh(artist_counts.index[::-1], artist_counts.values[::-1], color=colors)
plt.axvline(x=mean_val, color='gray', linestyle='--', linewidth=1.2, label=f'Mean ({mean_val:.1f})')
plt.xlabel("Number of Tracks in Top 2000")
plt.title("Top 15 Artists by Track Count in Spotify Top 2000")
plt.legend()
plt.tight_layout()
plt.show()
```

**Refinements**

**Change 1: Color encoding to highlight top artists**

I added a second color (orange) for the top 3 artists to visually distinguish outliers 
from the rest of the field. This directly supports the low level "detect outliers" task.
The gap between Taylor Swift/Drake/Morgan Wallen and the remaining artists is now 
immediately perceivable without requiring precise bar length comparison.

**Change 2: Mean reference line**

I added a vertical dashed line at the mean track count across the top 15. This provides 
context for what "typical" looks like among frequent artists, making the ranking task 
more meaningful. Viewers can now quickly identify who is above average vs. merely in 
the top 15.

Now that you have completed one visualization and gone through the design refinement process, repeat the same workflow for the remaining **three** visualizations in your plan. Your final submission should include **4 total visualizations**, each with its own reasoning and refinement notes.


```python
import pandas as pd
import matplotlib.pyplot as plt

spotify = pd.read_csv("spotify2000.csv")

top5 = spotify['artist_name'].value_counts().head(5).index.tolist()

def categorize(artist):
    if artist in top5:
        return artist
    return 'Others'

spotify['artist_group'] = spotify['artist_name'].apply(categorize)

grouped = spotify.groupby(['year', 'artist_group']).size().unstack(fill_value=0)

cols = ['Others'] + [c for c in grouped.columns if c != 'Others']
grouped = grouped[cols]

grouped.plot(kind='bar', stacked=True, figsize=(14, 6),
             colormap='tab10')
plt.xlabel("Year")
plt.ylabel("Number of Tracks")
plt.title("Top Track Counts per Year: Top 5 Artists vs. Others")
plt.legend(title="Artist", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

**Viz 2: Artist Dominance Over Time**

This visualization shows how the number of top tracks per year is distributed across 
artists, revealing whether certain artists dominated specific eras or whether the top 
charts are spread across many artists.

**Refinement 1: Color by top artist**

I highlighted Taylor Swift's bars in orange to make her cross-era dominance immediately 
visible. This supports the high level "identify trends" task, viewers can instantly 
see which years she contributed to rather than searching manually.

**Refinement 2: Stacked bar by top 5 artists vs. others**

Instead of a simple count per year, I stacked bars to show how much of each year's 
top tracks belong to the top 5 artists vs. everyone else. This better supports the 
"compare distributions" task by showing concentration of dominance over time.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

spotify = pd.read_csv("spotify2000.csv")

artist_counts = spotify['artist_name'].value_counts()
spotify['artist_type'] = spotify['artist_name'].apply(
    lambda x: 'Repeat Artist' if artist_counts[x] > 1 else 'One-Hit Wonder'
)

groups = [
    spotify[spotify['artist_type'] == 'One-Hit Wonder']['danceability'].dropna(),
    spotify[spotify['artist_type'] == 'Repeat Artist']['danceability'].dropna()
]
labels = ['One-Hit Wonder', 'Repeat Artist']

fig, ax = plt.subplots(figsize=(8, 6))
bp = ax.boxplot(groups, labels=labels, patch_artist=True,
                boxprops=dict(facecolor='steelblue', alpha=0.5))

for i, group in enumerate(groups):
    x = np.random.normal(i + 1, 0.05, size=len(group))
    ax.scatter(x, group, alpha=0.15, s=8, color='steelblue')

ax.set_ylabel("Danceability")
ax.set_title("Danceability: One-Hit Wonders vs. Repeat Artists")
plt.tight_layout()
plt.show()
```

**Viz 3: One Hit Wonders vs. Repeat Artists**

This visualization compares the danceability distributions of artists who appear only 
once in the top 2000 ("one-hit wonders") versus those who appear multiple times 
("repeat artists"), to detect whether repeat-charting artists share a distinct sonic profile.

**Refinement 1: Box plot instead of bar**

I used a box plot rather than a simple bar of means, because it shows the full 
distribution (median, spread, outliers) rather than just a single summary statistic. 
This better supports the "compare distributions" mid-level task.

**Refinement 2: Overlay individual data points**

I added a strip of individual points over the box plot (jittered) to show the actual 
density of values. This supports the low-level "detect outliers" task by making 
individual extreme values visible.

```python
import pandas as pd
import matplotlib.pyplot as plt

spotify = pd.read_csv("spotify2000.csv")

artist_counts = spotify['artist_name'].value_counts()
spotify['artist_type'] = spotify['artist_name'].apply(
    lambda x: 'Repeat Artist' if artist_counts[x] > 1 else 'One-Hit Wonder'
)

genre_props = (spotify.groupby(['artist_type', 'genre'])
               .size()
               .unstack(fill_value=0)
               .apply(lambda x: x / x.sum(), axis=1))

genre_props = genre_props.T
genre_props = genre_props.sort_values('Repeat Artist', ascending=True)

genre_props.plot(kind='barh', stacked=False, figsize=(10, 8),
                 color=['steelblue', '#e05c2a'], alpha=0.8)
plt.xlabel("Proportion of Tracks")
plt.title("Genre Distribution: One-Hit Wonders vs. Repeat Artists")
plt.legend(title="Artist Type")
plt.tight_layout()
plt.show()
```

**Viz 4: Genre Breakdown by Artist Type**

This visualization summarizes the genre composition of one-hit wonders versus repeat 
artists, to identify whether certain genres are more associated with sustained chart 
presence.

**Refinement 1: Normalized stacked bar (proportions)**

Instead of raw counts, I used proportions so the two groups are directly comparable 
despite having very different sizes. This better supports the "identify correlation" 
task — raw counts would be misleading since there are far more one-hit wonders than 
repeat artists.

**Refinement 2: Sorted genres by repeat artist proportion**

I sorted the genres by how frequently they appear among repeat artists, making it easy 
to see which genres dominate at the top. This supports the "rank" low level task and 
draws the viewer's eye to the most meaningful patterns.
