#### Extension Options After a Simple Visualization

##### 1) Change the Mark Type
- If we are on a line plot, can we have the combination of point + line or line + area (area plot)
- If we are on scatter plot, whether heatmap can be a better way to show the density of the dots
- If we are on a Box plot, can a violin plot be used to include the distribution
- If we are on a bar plot, considering the ways of sorting the bars:
    - Stacked or group
    - Sort by Alphabetic order
    - Sort by ranking values
    - Is there any particular features we want to focus?
    - Try a couple of different layout to compare
- If you are on a map, think about choropleth vs. symbol map. 

##### 2) Upgrade the Encoding Channels
- Review the channel table to see if the most effective channel has been used
- Is color necessary? What is the color map? Is it appropriate? 
- Whether at least one of the keys is on the position channel?
- How your task matches your design choice (precise number reading vs. pattern recognization)
- Have you conisder the possibility for parallel and radial axis design?

##### 3) Add Reference and Context for highlighting
- Mean/median reference lines
- Threshold lines (e.g., popularity > 80)
- Highlight top or bottom options. 
- Highlight outliers with different color saturation. 
- Regression line or smooth cureve if want to discuss the linear/nonline trend

##### 4) Improve Scale and Axis Choices
- Linear vs log scale
- Sorting categories
- Binning continuous variables
- Fixing misleading axis ranges
- Filtering if there are too many levels in categorical data

##### 5) Add Facet
