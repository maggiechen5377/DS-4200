### Part 2.1 — Understand the Scales (Lines 20–36)(10 points)
Read the code from **lines 20–36**. In this section, the chart defines **four scales**.

- Identify the **four scales** (by name and purpose).
- Explain why the chart uses **two x-scales**:
  - What does each x-scale represent?
  - How are the two x-scales different?
  - How do they work together to position the side-by-side bars?

You may paste the code snippet into an AI tool to translate it into another language you are familiar with (e.g., Python). Use the translation to help your understanding, but your explanation should be **in your own words**.

**What to submit (Type in the following markdown chunk):**
- A short answer (6–10 sentences) explaining the four scales and the role of the two x-scales.

The code defines four scales to position and color the bars in the chart. The first scale, x0, is a band scale that maps each car body style (such as convertible, sedan, or hatchback) to a horizontal position across the full width of the chart. The second scale, x1, is also a band scale but it maps each drive wheel type (FWD or RWD) to a position within the space that x0 allocates for each body style. The third scale, y, is a linear scale that maps car price values to vertical pixel positions, ranging from the maximum price at the top down to zero at the bottom. The fourth scale, color, is an ordinal scale that assigns a distinct color to each drive wheel type so they can be visually distinguished in the chart.
The chart uses two x-scales because it needs to display two bars side by side within each body style group. x0 handles the outer grouping by dividing the chart into sections for each body style, while x1 handles the inner grouping by splitting each section into slots for FWD and RWD. When drawing each bar, the code combines both scales: x0 gives the starting position of the group, and x1 gives the offset within that group. Together they make sure the two bars sit next to each other without overlapping.

The bar drawing works like a nested loop. The outer loop iterates over every row in the dataset and creates a group element positioned at the x0 location for that row's body style. The inner part of each group then draws a rectangle whose horizontal position is determined by x1 based on the drive wheel type, so FWD and RWD bars land side by side within the group. The height of each bar is calculated by subtracting the y scale value from the bottom of the chart. The legend is drawn separately by looping over the unique drive wheel categories and placing a colored square and a text label for each one.
