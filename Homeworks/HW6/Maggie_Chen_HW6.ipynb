### Part 1.1 Add selection (10 points)

Create a **scatter plot** showing the relationship between **average personal income** (x-axis) and **average health score** (y-axis).

- Encode **region** using **color**.
- Add a **tooltip** that displays the **country name** and **population**.
- Add an interaction that allows the user to **select a single country** on the visualization(this should be a point selection instead of any other widgets. ):
  - The selected country should be **highlighted**.
  - All other countries should appear in **light grey** when a selection is made.


```python
import altair as alt
import pandas as pd
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)

gapminder = pd.read_csv('gapminder-health-income.csv')

# Point selection on click
selection = alt.selection_point()

scatter = alt.Chart(gapminder).mark_point(filled=True, size=80).encode(
    x=alt.X('income:Q', title='Average Personal Income'),
    y=alt.Y('health:Q', title='Average Health Score'),
    color=alt.condition(
        selection,
        alt.Color('region:N', legend=alt.Legend(title='Region')),
        alt.value('lightgrey')),
    tooltip=['country:N', 'population:Q']
).add_params(
    selection
).properties(
    title='Income vs. Health by Country',
    width=600,
    height=400)

scatter
```

### Part 1.2 Customize the color (5 points)

Now choose a **customized color map** for the scatter plot from the previous question.

- Select your colors using an external resource such as **ColorBrewer** or **Colorgorical**.  
  **Do NOT** use any of the pre-defined color palettes from Matplotlib.
- Briefly **explain how and why** you chose this color map (e.g., perceptual uniformity, color-blind accessibility, clear separation between regions).
- Apply the selected color map to the plot.

You may use the following line to include your own color list (where `legend_color` is your list of selected colors):

    alt.Color('region:N', scale=alt.Scale(range=legend_color))

```python
import altair as alt
import pandas as pd
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)

gapminder = pd.read_csv('gapminder-health-income.csv')

legend_color = [
    '#1f78b4',  # bold blue       → america
    '#ff7f00',  # vivid orange    → east_asia_pacific
    '#6a3d9a',  # deep purple     → europe_central_asia
    '#e31a1c',  # bright red      → middle_east_north_africa
    '#33a02c',  # strong green    → south_asia
    '#ffff33',  # bright yellow   → sub_saharan_africa
]

selection = alt.selection_point()

scatter = alt.Chart(gapminder).mark_point(filled=True, size=80).encode(
    x=alt.X('income:Q', title='Average Personal Income'),
    y=alt.Y('health:Q', title='Average Health Score'),
    color=alt.condition(
        selection,
        alt.Color('region:N',
                  scale=alt.Scale(range=legend_color),
                  legend=alt.Legend(title='Region', labelLimit=200)),
        alt.value('lightgrey')),
    tooltip=['country:N', 'population:Q']
).add_params(
    selection
).properties(
    title='Income vs. Health by Country',
    width=600,
    height=400)

scatter
```

**Justification**
* Maximum visual distinction — the 6 colors span completely different hue families (blue, orange, purple, red, green, yellow), making each region immediately identifiable without careful inspection.
* Colorblind accessibility — the palette avoids problematic red/green pairings as the primary distinguishing colors, and the high contrast between hues ensures regions remain distinguishable across common color vision deficiencies.
* Categorical suitability — the colors carry no implied ordering or magnitude, which is appropriate since region is an unordered nominal variable.

### Part 1.3 Select across multiple panels (5 points)

Instead of using a point (single-country) selection, add an **interval selection** to the scatter plot in **Part 1.1**.

- Create an **interval brush** that allows the user to select an **income range only** (i.e., the selection should operate along the **x-axis**).
- Using this selection, generate a **second plot** to the right that updates dynamically to show the relationship between **income** and **population** **for the countries within the selected income range**.


```python
import altair as alt
import pandas as pd
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)

gapminder = pd.read_csv('gapminder-health-income.csv')

# Interval brush restricted to x-axis only
brush = alt.selection_interval(encodings=['x'])

# Left plot: income vs health (from Part 1.1, with interval brush)
scatter = alt.Chart(gapminder).mark_point(filled=True, size=80).encode(
    x=alt.X('income:Q', title='Average Personal Income'),
    y=alt.Y('health:Q', title='Average Health Score'),
    color=alt.condition(
        brush,
        alt.Color('region:N', legend=alt.Legend(title='Region')),
        alt.value('lightgrey')),
    tooltip=['country:N', 'population:Q']
).add_params(
    brush
).properties(
    title='Income vs. Health by Country',
    width=400,
    height=400)

# Right plot: income vs population, filtered by brush selection
pop_plot = alt.Chart(gapminder).mark_point(filled=True, size=80).encode(
    x=alt.X('income:Q', title='Average Personal Income'),
    y=alt.Y('population:Q', title='Population'),
    color=alt.Color('region:N', legend=alt.Legend(title='Region')),
    tooltip=['country:N', 'population:Q']
).transform_filter(
    brush
).properties(
    title='Income vs. Population (Selected Range)',
    width=400,
    height=400)

chart = scatter | pop_plot
chart
```

### Part 1.5 Data binding and widgets (10 points)

Modify the **Part 1.1** scatter plot by replacing the point-clicked interaction with a **radio-button control** for **region**.

- Add a **radio button** that allows the user to select **one region at a time**.
- When a region is selected, **highlight only the points** from that region and make all other points **grey**.
- Also encode the selection using **opacity**:
  - Points in the selected region should have **opacity = 1**
  - Points in all other regions should have **opacity = 0.1**

```python
import altair as alt
import pandas as pd
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)

gapminder = pd.read_csv('gapminder-health-income.csv')

# Radio button bound to region field
selection = alt.selection_point(
    fields=['region'],
    bind=alt.binding_radio(
        options=sorted(gapminder['region'].unique().tolist()),
        name='Region: '))

scatter = alt.Chart(gapminder).mark_point(filled=True, size=80).encode(
    x=alt.X('income:Q', title='Average Personal Income'),
    y=alt.Y('health:Q', title='Average Health Score'),
    color=alt.condition(
        selection,
        alt.Color('region:N', legend=alt.Legend(title='Region')),
        alt.value('lightgrey')),
    opacity=alt.condition(
        selection,
        alt.value(1),
        alt.value(0.1)),
    tooltip=['country:N', 'population:Q']
).add_params(
    selection
).properties(
    title='Income vs. Health by Country',
    width=600,
    height=400)

scatter
```

### Part 1.6 Add filter with bars (10 points)

Modify the **Part 1.1** scatter plot by adding a **slider control** for population.

- Add a **slider bar** that allows the user to set a **population cutoff value**.
- Filter the data dynamically so that **only countries with population less than the selected value** are displayed.
- Set the **initial cutoff value** of the slider to **1,000,000**.

```python
import altair as alt
import pandas as pd
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)

gapminder = pd.read_csv('gapminder-health-income.csv')

# Slider bound to a parameter for population cutoff, initial value 1,000,000
pop_slider = alt.binding_range(
    min=0,
    max=int(gapminder['population'].max()),
    step=1000000,
    name='Population Cutoff: ')

pop_cutoff = alt.param(bind=pop_slider, value=1000000)

scatter = alt.Chart(gapminder).mark_point(filled=True, size=80).encode(
    x=alt.X('income:Q', title='Average Personal Income'),
    y=alt.Y('health:Q', title='Average Health Score'),
    color=alt.Color('region:N', legend=alt.Legend(title='Region')),
    tooltip=['country:N', 'population:Q']
).transform_filter(
    alt.datum.population < pop_cutoff
).add_params(
    pop_cutoff
).properties(
    title='Income vs. Health by Country (Population Filter)',
    width=600,
    height=400)

scatter
```
