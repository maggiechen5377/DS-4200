Bar Plot Mental Tasks
* Discrete Comparison: The viewer is meant to look at a specific category (ex "Execution") and compare the heights of the bars to see which gymnast scored highest in that specific area.
* Lookup: It supports the task of finding a specific value for a specific entity (ex What was Gymnast 2’s score in Difficulty?).
* Ranking: Determining the order of performance within a single category (ex ranking 1st, 2nd, and 3rd for landing).

Radar Plot Mental Tasks
* Profile/Shape Recognition: The viewer looks at the overall shape created by a gymnast's scores across all categories. For example, a large shape indicates high performance across the board, while a skewed shape reveals specific strengths and weaknesses.
* Holistic Evaluation: Rather than comparing individual bars, the brain quickly assesses the balance or symmetry of the athletes' skills.
* Pattern Matching: Comparing the overlap of two shapes to see if two athletes have similar skill profiles (ex are they both artistic gymnasts or power gymnasts?).

Additional Visualization 
* Proposal: Stacked Bar Chart
    * While the current grouped bar chart is great for comparing individual categories, it makes it very difficult for the viewer to calculate the overall winner. A stacked bar chart would place the scores for Execution, Landing, Style, etc., on top of one another for each gymnast. This allows the viewer to mentally assess the total length of the bar to determine the overall score while still seeing how each category contributed to that total.
